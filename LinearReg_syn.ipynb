```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Generate synthetic data: y = 2.5 * x + 1.5 + noise
np.random.seed(42)
X = np.random.rand(100, 1) * 10
noise = np.random.randn(100, 1) * 2
y = 2.5 * X + 1.5 + noise

# 2. Fit the linear regression model
model = LinearRegression()
model.fit(X, y)

# 3. Retrieve slope (coefficient) and intercept
slope = model.coef_[0][0]
intercept = model.intercept_[0]

print(f"Estimated Slope (True = 2.5): {slope:.4f}")
print(f"Estimated Intercept (True = 1.5): {intercept:.4f}")
```

```python
# 4. Plot the results
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', alpha=0.6, label='Synthetic Data')
plt.plot(X, model.predict(X), color='red', linewidth=2, label=f'Fit Line: y = {slope:.2f}x + {intercept:.2f}')
plt.title('Linear Regression with Synthetic Data')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
