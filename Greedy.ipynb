```python
def first_fit(items, bin_capacity):
    # This list will store the remaining capacity of each active bin
    bin_rem_capacities = []
    # This list will keep track of which items are placed in which bin
    bins = []

    for item in items:
        if item > bin_capacity:
            raise ValueError(f"Item size {item} exceeds maximum bin capacity {bin_capacity}")

        # Find the first bin that can accommodate the current item
        placed = False
        for i in range(len(bin_rem_capacities)):
            if bin_rem_capacities[i] >= item:
                bin_rem_capacities[i] -= item
                bins[i].append(item)
                placed = True
                break

        # If no bin could accommodate the item, start a new bin
        if not placed:
            bin_rem_capacities.append(bin_capacity - item)
            bins.append([item])

    return bins, len(bins)
```

```python
# Example usage:
items = [2, 5, 4, 7, 1, 3, 8]
bin_capacity = 10

packed_bins, num_bins = first_fit(items, bin_capacity)

print(f"Items to pack: {items}")
print(f"Bin Capacity: {bin_capacity}\n")
print(f"Number of bins required: {num_bins}")
for idx, contents in enumerate(packed_bins):
    print(f"Bin {idx + 1}: {contents} (Used space: {sum(contents)})")
```
